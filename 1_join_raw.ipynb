```python
import pandas as pd
import glob
import os

# Buscamos los archivos en la carpeta de datos
ruta = "data/BD *.xlsx"
archivos = glob.glob(ruta)
archivos.sort()

df_list = []
print("Iniciando Script 1: Fusión de datos RAW...")

for archivo in archivos:
    xls = pd.ExcelFile(archivo)
    for hoja in xls.sheet_names:
        # Leemos cada pestaña
        df = pd.read_excel(xls, sheet_name=hoja)
        df['Estacion'] = hoja 
        
        # Corrección de la columna de tiempo según el año
        if 'Fecha y hora' in df.columns:
            df.rename(columns={'Fecha y hora': 'Date'}, inplace=True)
        elif 'date' in df.columns:
            df.rename(columns={'date': 'Date'}, inplace=True)
            
        df_list.append(df)
        print(f"✔️ {os.path.basename(archivo)} - Hoja: {hoja} extraída.")

# Unimos la historia completa en un solo dataframe
df_raw = pd.concat(df_list, ignore_index=True)

# Aseguramos el formato de fecha y ordenamos lógicamente
df_raw['Date'] = pd.to_datetime(df_raw['Date'])
df_raw.sort_values(by=['Estacion', 'Date'], inplace=True)
df_raw.reset_index(drop=True, inplace=True)

# Creamos la carpeta 'raw' si no existe
os.makedirs("data/raw", exist_ok=True)

# Exportamos en ambos formatos
ruta_parquet = "data/raw/sima_raw.parquet"
ruta_csv = "data/raw/sima_raw.csv"

df_raw.to_parquet(ruta_parquet, index=False)
print(f"\n📦 Archivo Parquet guardado en: {ruta_parquet}")

df_raw.to_csv(ruta_csv, index=False)
print(f"📄 Archivo CSV guardado en: {ruta_csv}")

print(f"\n✅ RAW Dataset listo. Total de filas: {df_raw.shape[0]}. Las banderas y letras se mantienen intactas.")
```
